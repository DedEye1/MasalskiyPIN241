### 5. Вычислите  для  столбца  fnlwgt  среднее  значение,  моду,  медиану, дисперсию и среднеквадратичное отклонение

```python
mean = df['fnlwgt'].mean()
mode = df['fnlwgt'].mode()[0]
median = df['fnlwgt'].median()
var = df['fnlwgt'].var()
std = df['fnlwgt'].std()

print(f'Среднее значение: {mean}', f'Мода: {mode}', f'Медиана: {median}', \
      f'Дисперсия: {var}', f'Среднеквадратичное отклонение: {std}', sep=';\n')
```

### 7.  С помощью groupby() и agg() определите

* средний возраст и медианный возраст по полу, расе или семейному положению

```python
age_sex = df.groupby('sex').agg({'age': ['mean', 'median']})
display(age_sex)

age_race = df.groupby('race').agg({'age': ['mean', 'median']})
display(age_race)

age_marital = df.groupby('marital-status').agg({'age': ['mean', 'median']})
display(age_marital)
```

* распределение дохода по полу. Какой процент мужчин и женщин имеют доход >50K?

```python
salary_sex = df.groupby('sex').agg({'salary': lambda x: (x == '>50K').mean() * 100})

round(salary_sex, 2)
```

* среднее  количество  рабочих  часов  в  неделю  по  роду  занятий  или семейному положению

```python
av_h_occupation = df.groupby('occupation').agg({'hours-per-week': 'mean'})
display(av_h_occupation)

av_h_marital = df.groupby('marital-status').agg({'hours-per-week': 'mean'})
display(av_h_marital)
```

* количество людей в каждом классе работы и для каждой профессии

```python
count = df.groupby(['workclass', 'occupation']).size()
count_df = pd.DataFrame(count).rename(columns={0: 'Count'})

count_df
```

* самые  распространенные  профессии  (occupation)  для  мужчин  и женщин в каждой расе (race)

```python
common_occupation = df.groupby(['race', 'sex']).agg({'occupation': lambda x: x.mode()})

common_occupation
```

### 8.  С помощью pivot_table() постройте сводные таблицы, показывающие

* влияние  возраста  и  образования  на  доход.  Разбить  возраст  на группы (bins) и посмотреть долю высокого дохода в каждой группе для разных уровней образования

```python
df['age_group'] = pd.cut(df['age'], bins=[0, 20, 50, 100], labels=['Young', 'Adult', 'Senior'])

piv = df.pivot_table(
    index='education',
    columns='age_group',
    values='salary',
    aggfunc=lambda x: (x == '>50K').mean() * 100,
    observed=False
)

round(piv, 2)
```

* вероятность  высокого  дохода  в  зависимости  от  профессии  и семейного положения. Где «счастливые» комбинации?

```python
piv = df.pivot_table(
    index='occupation',
    columns='marital-status',
    values='salary',
    aggfunc=lambda x: (x == '>50K').mean() * 100
)

round(piv, 2)
```

* взаимосвязь  образования,  пола  и  дохода.  Средний  возраст  или процент высокого дохода (>50K) в разрезе education и sex

```python
piv = df.pivot_table(
    index='education',
    columns='sex',
    values=['age', 'salary'],
    aggfunc={
        'age': 'mean',
        'salary': lambda x: (x == '>50K').mean() * 100
    }
).rename(columns={
    'age': 'Average Age',
    'salary': 'High Salary (%)'
})

round(piv, 2)
```

* среднее количество рабочих часов в неделю для людей с высоким и низким доходом в разрезе типа занятости (workclass)

```python
piv = df.pivot_table(
    index='workclass',
    columns='salary',
    values='hours-per-week',
    aggfunc='mean'
)

round(piv, 2)
```

### 9.  Создайте  новый  признак  is_married,  который  упрощает  имеющийся признак marital-status

Разделение на категории:

```python
conditions = [
    df['marital-status'].isin([' Divorced', ' Never-married', ' Widowed', ' Separated']),
    df['marital-status'].isin([' Married-AF-spouse', ' Married-civ-spouse', ' Married-spouse-absent'])
]
choices = [0, 1]

df['is_married'] = np.select(conditions, choices)
df.head(10)
```
